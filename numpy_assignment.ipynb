```python
# 5
import numpy as np
np.random.seed(10)
scores = np.random.randint(20, 101, 12)
print("Original Scores:", scores)
status = np.where(scores >= 40, "Pass", "Fail")
print("Student Status:", status)
updated_scores = np.where(scores < 40, scores + 5, scores)
print("Updated Scores:", updated_scores)
```

```python
# 6

import numpy as np

np.random.seed(10)
marks = np.random.randint(50, 101, (6, 4))

print("Marks Matrix:")
print(marks)

avg = np.mean(marks, axis=1)
print("Average Marks:", np.round(avg, 2))

highest = np.max(marks, axis=0)
print("Highest Marks in Each Subject:", highest)

result = marks[avg > 75]
print("Students scoring average above 75:")
print(result)
```

```python
# 7

import numpy as np

houses = np.array([
    [1200, 2, 55],
    [1800, 3, 85],
    [950, 1, 40],
    [2400, 4, 120],
    [1500, 3, 70]
])

minimum = np.min(houses, axis=0)
maximum = np.max(houses, axis=0)

normalized = (houses - minimum) / (maximum - minimum)

print("Normalized Data:")
print(np.round(normalized, 3))
```

```python
# 8
morning_sales = np.array([
    [15, 20],
    [10, 12],
    [25, 30]
])

evening_sales = np.array([
    [18, 22],
    [14, 15],
    [20, 28]
])
full_sales = np.concatenate((morning_sales, evening_sales), axis=1)

print(full_sales)
```

```python
branch4 = np.array([[12, 16, 15, 19]])

final_sales = np.vstack((full_sales, branch4))

print("Final Sales:")
print(final_sales)
```

```python
total = np.sum(final_sales, axis=0)
print("Total Sales:", total)
```

```python
# 9
ratings = np.array([5, 4, 3, 5, 2, 5, 4, 1, 3, 5,4, 4, 2, 5, 3, 4, 5, 1, 4, 3])
unique, count = np.unique(ratings, return_counts=True)

percentage = (count / len(ratings)) * 100

print("Rating Summary:")

for i in range(len(unique)):
    print(f"Rating {unique[i]}: {count[i]} responses ({percentage[i]}%)")

```
