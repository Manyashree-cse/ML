```python
import numpy as np
import pandas as pd
import matplotlib .pyplot as plt
import seaborn as sns

df=pd.read_csv('/content/Iris.csv')
```

```python
# boxplot
sns.boxplot(x=df['SepalLengthCm'])
plt.show()

# lineplot
sns.lineplot(x=df['SepalLengthCm'],y=df['SepalWidthCm'])
plt.show()

# scatterplot
sns.scatterplot(x=df['SepalLengthCm'],y=df['SepalWidthCm'])
plt.show()

# barplot
sns.barplot(x=df['SepalLengthCm'],y=df['SepalWidthCm'])
plt.show()

# histplot
sns.histplot(x=df['SepalLengthCm'],y=df['SepalWidthCm'])
plt.show()
```

```python
# boxplot
plt.boxplot(x=df['SepalLengthCm'])
plt.show()

# lineplot
plt.plot(df['SepalLengthCm'],df['SepalWidthCm'])
plt.show()

# scatterplot
plt.scatter(x=df['SepalLengthCm'],y=df['SepalWidthCm'])
plt.show()

# barplot
plt.bar(df['SepalLengthCm'],df['SepalWidthCm'])
plt.show()

```

```python
# histplot
plt.hist2d(df['SepalLengthCm'],df['SepalWidthCm'])
plt.show()
```

```python
plt.plot(df['SepalLengthCm'],df['SepalWidthCm'])
plt.xlabel('SepalLengthCm')
plt.ylabel('SepalWidthCm')
plt.title('SepalLengthCm vs SepalWidthCm')
plt.show()
```

```python
species_count = df['Species'].value_counts()
print(species_count)
# sns.set_theme(style="whitegrid")

plt.pie(
    species_count,
    labels=species_count.index,
    autopct='%1.1f%%',
    colors=sns.color_palette('pastel'),
    explode=[0.1, 0, 0],
    shadow=True
)

plt.title("Iris Species Distribution")
plt.show()
```

```python
sns.countplot(data=df,x='Species')
plt.show()
```
