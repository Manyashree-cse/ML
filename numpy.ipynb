```python
a = np.array([[1, 2, 3],
              [4, 5, 6]])

for i in a:
    print(i)
```

```python
a = np.array([[1, 2, 3],
              [4, 5, 6]])

for i in a:
    for j in i:
        print(j)
```

```python
a = np.array([[1, 2, 3],
              [4, 5, 6]])

for i in np.nditer(a):
    print(i)
```
